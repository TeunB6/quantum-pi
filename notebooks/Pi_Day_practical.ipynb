How can we improve this series? We can apply convergence accelerating transformations. An example of which is the so-called [Shanks transformation](https://en.wikipedia.org/wiki/Shanks_transformation). This transformation speeds up the convergence of converging infinite series. Since the infinite series we are currently using converges to $\pi$, applying this transformation will speed convergence up drastically. A cool thing about this transformation is that we can apply it to itself, improving convergence further.

\end{equation}

### Optimizing
The code we are using currently is constantly recalculating values it has already calculated. As at each depth we call the shanks transformation of the previous n again, this takes way too long. A way to fix this would be, naturally, to store the values we have already computed. This can be done manually using a dictionary(hashmap). However, as we are using python someone has implemented this idea in a much better way. In this case we can use the lru_cache decorator from the functools library. Use this, or any other strategy, to optimize the code below.

## Gauss-Legendre Algorithm
The Algorithms you have seen so far were all invented before the 18th century. The algorithm presented here is a bit more modern. The algorithm is, in simple terms, based on finding a mean of a distribution centered on $\pi$. The main strength of this algorithm is its fast convergence. Where the Leibniz series has a sub-linear convergence, the Gauss-Legendre Algorithm has quadratic convergence. This means that each iteration of the algorithm doubles the number of correct digits.

The algorithm works by iteratively calculating the arithmetic and geometric mean of two numbers (seen in 2.1 and 2.2 respectively). The starting values of these numbers are specified as $a_0$ and $b_0$ in 1. The limit of the the arithmetic-geometric mean of $a_0$ and $b_0$ as calculated in step 2, can be used in the formula of step 3 to approximate $\pi$. The proof of this algorithm is quite complicated and is not outlined here, if you are interested it can be found [here](https://arxiv.org/abs/1907.04110). The authors use a slightly different formula for 3 but the idea is the same.

1. Setting the initial values

   t_{n+1} &= t_n - p_n(a_n-a_{n_1})^2 \\
   p_{n+1} &= 2p_n
   \end{split}
3. Using the values to approximate pi:
\begin{equation}
\pi ≈ \frac{(a_{n+1} + b_{n+1})^2}{4t_{n+1}}
\end{equation}
